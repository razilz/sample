### check prime numbers

```python
num = int(input("enter the number: "))

if num >1:
    for i in range(2,num):
        if num % i == 0:
            print("not prime")
            break
    else:
        print("prime")

else:
    print("not prime")
```

### Student Grade Calculator

```python
# student grade calculator

# input marks
subject1 = float(input("enter mark for subject 1: "))
subject2 = float(input("enter mark for subject 2: "))
subject3 = float(input("enter mark for subject 3: "))

# calculate total and avarage
total = subject1 + subject2 + subject3
avarage = total / 3 

# Display result 
print("\nTotal Marks:", total)
print("Avarage Marks:",avarage)

# Grade calculation
if avarage >= 90:
    grade = "A+"
elif avarage >= 75:
    grade = "A"
elif avarage >= 60:
    grade = "B"
elif avarage >= 50:
    grade = "C" 
else:
    grade = "fail"

print("Grade:", grade)           
```

### Reverse Triangle

```python
rows = 11 
for i in range (rows,0,-1):
    print(" #"*i)
```

### Diomond TRiangle

```python
rows = 5
for i in range (1, rows + 1 ):
    print (" " * (rows - i)+ "# " * i)


for i in range (rows - 1,0,-1):
    print(" " *(rows - i)+ "# " * i)


```

## Square Pattern

```python
n = 5
for i in range(n):
    for j in range(n):
        print("*", end="")
    print()    
```

### rigth tringle

```python
n=5
for i in range(1, n+1):
    for j in range(i):
        print("*",end="")
    print()    
```

### inverted 

```python
n=5 
for i in range(n-1,0,-1):
    for j in range(i):
        print("*",end="")
    print()    
```

### Quiz Game Code

```python
print("welcome to quiz game!\n")

score = 0

# Q1
q1 = input("1. what is the capital of india? ")
if q1.lower() == "delhi":
    print(" correct!")
    score +=1
else:
    print("wrong! Answer: delhi")

# Q2
q2 = input("2. 5 + 3 +? ")
if q2 == "8":
    print("correct! ")
    score +=1
else:
    print("wrong! Answer: 8 ")

# Q3
q3 = input("3. which language is used for AI? ")
if q3.lower()== "python":
    print("correct!")
    score += 1
else:
    print("wrong! Answer: python")


# final score
print("\n your score:" ,score,"/3")

if score ==3 :
    print("excellent")
elif score == 2:
    print("good job!")
else:
    print("keep practicing")      
    
```
